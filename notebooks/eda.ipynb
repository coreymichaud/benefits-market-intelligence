## Libraries

```python
# Libraries
import pandas as pd
from innovation_summit.config import COMBINED_DATA_PATH
```

```python
# Displaying all
pd.set_option("display.max_rows", None)
```

## Load data

```python
# Loading data
F5500 = pd.read_parquet(COMBINED_DATA_PATH / "F5500.parquet")
SCH_A = pd.read_parquet(COMBINED_DATA_PATH / "SCH_A.parquet")
SCH_C_P1_I2 = pd.read_parquet(COMBINED_DATA_PATH / "SCH_C_P1_I2.parquet")
```

## Shape of data

```python
# Shape
print(f"F5500 has {F5500.shape[0]} rows & {F5500.shape[1]} columns")
print(f"SCH_A has {SCH_A.shape[0]} rows & {SCH_A.shape[1]} columns")
print(f"SCH_C_P1_I2 has {SCH_C_P1_I2.shape[0]} rows & {SCH_C_P1_I2.shape[1]} columns")
```

## Data types

```python
# Data type counts
print(f"F5500 has the dtypes counts of:\n{F5500.dtypes.value_counts()}\n")
print(f"SCH_A has the dtypes counts of:\n{SCH_A.dtypes.value_counts()}\n")
print(f"SCH_C_P1_I2 has the dtypes counts of:\n{SCH_C_P1_I2.dtypes.value_counts()}\n")
```

## Any missing data?

```python
# No missing data
print(f"F5500 columns with no missing data:\n{F5500.isna().sum()[F5500.isna().sum() == 0]}\n")
print(f"SCH_A columns with no missing data:\n{SCH_A.isna().sum()[SCH_A.isna().sum() == 0]}\n")
print(f"SCH_C_P1_I2 columns with no missing data:\n{SCH_C_P1_I2.isna().sum()[SCH_C_P1_I2.isna().sum() == 0]}")
```

```python
# F5500 missing data
pd.DataFrame({
    "missing_count": F5500.isna().sum(),
    "missing_percent": F5500.isna().mean().mul(100)
}).query("missing_count > 0").sort_values("missing_percent", ascending=False)
```

```python
# SCH_A missing data
pd.DataFrame({
    "missing_count": SCH_A.isna().sum(),
    "missing_percent": SCH_A.isna().mean().mul(100)
}).query("missing_count > 0").sort_values("missing_percent", ascending=False)
```

```python
# SCH_C_P1_I2 missing data
pd.DataFrame({
    "missing_count": SCH_C_P1_I2.isna().sum(),
    "missing_percent": SCH_C_P1_I2.isna().mean().mul(100)
}).query("missing_count > 0").sort_values("missing_percent", ascending=False)
```
